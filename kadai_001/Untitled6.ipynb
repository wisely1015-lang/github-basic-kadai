```python
import requests
from bs4 import BeautifulSoup

# 取得したいYahoo!ニュースの記事URL
url = "https://news.yahoo.co.jp/articles/562a67962b59f7022beca5ef9166e74c9698ab67"

# ブラウザからのアクセスに近づけるためUser-Agentを設定
headers = {
    "User-Agent": "Mozilla/5.0"
}

# Webページを取得
response = requests.get(url, headers=headers)

# 文字化け対策
response.encoding = response.apparent_encoding

# HTMLを解析
soup = BeautifulSoup(response.text, "html.parser")

# 記事本文を取得
article_body = soup.find("div", class_="article_body")

if article_body:
    print(article_body.get_text("\n", strip=True))
else:
    print("記事本文を取得できませんでした。")
```
