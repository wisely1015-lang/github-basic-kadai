```python
from openpyxl import Workbook
from openpyxl.styles import Border, Side, Alignment, Font
from datetime import datetime

# ----------------------------
# 1. 新しいExcelファイルを作成
# ----------------------------
wb = Workbook()

# アクティブなシートを取得
ws = wb.active
ws.title = "請求書"


# ----------------------------
# 2. 請求書の基本情報を入力
# ----------------------------

ws["B2"] = "請求書"
ws["B2"].font = Font(bold=True, size=14)

ws["B4"] = "株式会社ABC"
ws["B5"] = "〒101-0022 東京都千代田区神田練塀町300"
ws["B6"] = "TEL:03-1234-5678 FAX:03-1234-5678"
ws["B7"] = "担当者名:鈴木一郎 様"

# No.
ws["F4"] = "No."
ws["G4"] = "0001"

# 日付
ws["F5"] = "日付"
ws["G5"] = datetime.now()

# YYYY/MM/DD形式に設定
ws["G5"].number_format = "yyyy/mm/dd"


# ----------------------------
# 3. 商品表を作成
# ----------------------------

ws["B10"] = "商品名"
ws["C10"] = "数量"
ws["D10"] = "単価"
ws["E10"] = "金額"

ws["B11"] = "商品A"
ws["C11"] = 2
ws["D11"] = 10000
ws["E11"] = "=C11*D11"

ws["B12"] = "商品B"
ws["C12"] = 1
ws["D12"] = 15000
ws["E12"] = "=C12*D12"

# 商品金額合計
ws["E13"] = "=SUM(E11:E12)"


# ----------------------------
# 4. 合計・消費税・税込合計
# ----------------------------

ws["B15"] = "合計"
ws["E15"] = "=SUM(E11:E12)"

ws["B16"] = "消費税"
ws["E16"] = "=E15*10%"

ws["B17"] = "税込合計"
ws["E17"] = "=E15+E16"


# ----------------------------
# 5. 罫線を設定
# ----------------------------

thin = Side(style="thin", color="000000")
border = Border(
    left=thin,
    right=thin,
    top=thin,
    bottom=thin
)

# 商品表
for row in ws.iter_rows(min_row=10, max_row=13, min_col=2, max_col=5):
    for cell in row:
        cell.border = border

# 合計欄
for row in ws.iter_rows(min_row=15, max_row=17, min_col=2, max_col=5):
    for cell in row:
        cell.border = border


# ----------------------------
# 6. 文字位置を調整
# ----------------------------

for cell in ["C10", "D10", "E10"]:
    ws[cell].alignment = Alignment(horizontal="center")

for row in range(11, 18):
    ws[f"C{row}"].alignment = Alignment(horizontal="right")
    ws[f"D{row}"].alignment = Alignment(horizontal="right")
    ws[f"E{row}"].alignment = Alignment(horizontal="right")


# ----------------------------
# 7. 列幅を調整
# ----------------------------

ws.column_dimensions["A"].width = 5
ws.column_dimensions["B"].width = 25
ws.column_dimensions["C"].width = 10
ws.column_dimensions["D"].width = 12
ws.column_dimensions["E"].width = 12
ws.column_dimensions["F"].width = 10
ws.column_dimensions["G"].width = 14


# ----------------------------
# 8. ファイル名に現在の日付を入れて保存
# ----------------------------

today = datetime.now().strftime("%Y%m%d")

file_name = f"請求書_{today}.xlsx"

wb.save(file_name)

print(f"{file_name} を作成しました。")
```
