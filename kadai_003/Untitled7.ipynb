```python
import pandas as pd
from openpyxl import load_workbook
from openpyxl.styles import PatternFill

# ----------------------------
# 1. Excelファイルを読み込む
# ----------------------------

df_2022 = pd.read_excel("2022_年間売上表.xlsx", sheet_name="Sheet1")
df_2023 = pd.read_excel("2023_年間売上表.xlsx", sheet_name="Sheet1")


# ----------------------------
# 2. 2つのデータフレームを連結する
# ----------------------------

df_all = pd.concat(
    [df_2022, df_2023],
    ignore_index=True
)


# ----------------------------
# 3. 商品・売上年ごとに集計する
# ----------------------------

df_summary = (
    df_all
    .groupby(["商品", "売上年"], as_index=False)["金額（千円）"]
    .sum()
)


# ----------------------------
# 4. Excelファイルに書き出す
# ----------------------------

file_name = "売上集計表.xlsx"

df_summary.to_excel(
    file_name,
    index=False
)


# ----------------------------
# 5. openpyxlでExcelを開く
# ----------------------------

wb = load_workbook(file_name)

ws = wb.active


# ----------------------------
# 6. ヘッダーを薄いグレーにする
# ----------------------------

gray_fill = PatternFill(
    fill_type="solid",
    fgColor="F2F2F2"
)

for cell in ws[1]:
    cell.fill = gray_fill


# ----------------------------
# 7. 保存する
# ----------------------------

wb.save(file_name)

print("売上集計表.xlsxを作成しました。")
```
