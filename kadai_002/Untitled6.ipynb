```python
import requests
from getpass import getpass

# 1. APIキーを入力
api_key = getpass("Google Maps APIキーを入力してください：")

# 2. 駅名を入力
station = input("駅名を入力してください：")

# 3. 駅名から緯度・経度を取得
geocode_url = "https://maps.googleapis.com/maps/api/geocode/json"

geocode_params = {
    "address": station,
    "key": api_key
}

geocode_response = requests.get(geocode_url, params=geocode_params)
geocode_data = geocode_response.json()

# 緯度・経度を取り出す
location = geocode_data["results"][0]["geometry"]["location"]

latitude = location["lat"]
longitude = location["lng"]

print("緯度：", latitude)
print("経度：", longitude)

# 4. 駅の周辺500mのレストランを検索
places_url = "https://maps.googleapis.com/maps/api/place/nearbysearch/json"

places_params = {
    "location": f"{latitude},{longitude}",
    "language": "ja",
    "radius": 500,
    "type": "restaurant",
    "key": api_key
}

places_response = requests.get(places_url, params=places_params)
places_data = places_response.json()

# 5. レストラン情報を表示
for restaurant in places_data["results"]:
    print("レストラン名：", restaurant.get("name", "情報なし"))
    print("評価：", restaurant.get("rating", "情報なし"))
    print("住所：", restaurant.get("vicinity", "情報なし"))
    print("------------------------------")
```
