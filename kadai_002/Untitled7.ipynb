```python
import pandas as pd

# ----------------------------
# 1. 元データを作成
# ----------------------------

data = {
    "日付": [
        "2023-05-17",
        "2023-05-18",
        "2023-05-19",
        "2023-05-20",
        "2023-05-21"
    ],
    "社員名": [
        "山田",
        "佐藤",
        "鈴木",
        "田中",
        "高橋"
    ],
    "売上": [
        100,
        200,
        150,
        300,
        250
    ],
    "部門": [
        "メーカー",
        "代理店",
        "メーカー",
        "商社",
        "代理店"
    ]
}

# DataFrameを作成
df = pd.DataFrame(data)


# ----------------------------
# 2. 平均売上を計算
# ----------------------------

average_sales = df["売上"].mean()

# 「平均売上」列を追加
df["平均売上"] = average_sales


# ----------------------------
# 3. 業績ランクを判定する関数
# ----------------------------

def performance(sales):
    if sales >= average_sales + 50:
        return "A"
    elif sales >= average_sales:
        return "B"
    else:
        return "C"


# performance関数を売上列に適用
df["業績ランク"] = df["売上"].apply(performance)


# ----------------------------
# 4. Excelファイルに書き出す
# ----------------------------

df.to_excel("業績.xlsx", index=False)

print("業績.xlsxを作成しました。")
```
